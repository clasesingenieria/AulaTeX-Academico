# 04 — Reward model

El reward model es un **juez de calidad**, no el generador de actividades. Primero se ejecuta una prueba de humo y luego, solo si el corpus tiene señal suficiente, el entrenamiento real. No se promueve con Spearman menor que 0.5.

```python
from pathlib import Path
import json
import subprocess
import sys

ROOT = Path.cwd()
if ROOT.name == "notebooks":
    ROOT = ROOT.parent.parent
elif ROOT.name == "aulaTEX":
    ROOT = ROOT.parent
package = "aulaTEX" if (ROOT / "aulaTEX/aulatex_training").is_dir() else "scripts"
data_root = ROOT / "aulaTEX/data" if (ROOT / "aulaTEX/data").is_dir() else ROOT / "data"
RUN_REAL_REWARD_TRAINING = False
command = [sys.executable, str(ROOT / package / "aulatex_training/train_reward_model.py")]
if RUN_REAL_REWARD_TRAINING:
    command += ["--train-file", str(data_root / "private/processed/reward.jsonl"), "--output-dir", str(ROOT / "aulaTEX/models/local/reward-model")]
else:
    command += ["--smoke-test", "--output-dir", str(ROOT / "aulaTEX/models/local/reward-smoke")]
print("Ejecutando:", command)
result = subprocess.run(command, cwd=ROOT, text=True)
print("Código de salida:", result.returncode)
```
