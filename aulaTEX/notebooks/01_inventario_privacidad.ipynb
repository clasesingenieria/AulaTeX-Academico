# 01 — Inventario y privacidad

Inventaría fuentes AulaTeX y busca datos sensibles antes de construir el corpus. No imprime el contenido documental completo. La revisión humana de licencias y privacidad sigue siendo obligatoria.

```python
from pathlib import Path
import sys
from collections import Counter

ROOT = Path.cwd()
if ROOT.name == "notebooks":
    ROOT = ROOT.parent.parent
elif ROOT.name == "aulaTEX":
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT))
if (ROOT / "aulaTEX/aulatex_training").is_dir():
    from aulaTEX.aulatex_training.motor_training import inventory_sources, privacy_findings, read_utf8
else:
    from scripts.aulatex_training.motor_training import inventory_sources, privacy_findings, read_utf8

inventory = inventory_sources(ROOT)
print({name: len(paths) for name, paths in inventory.items()})
findings = []
for path in inventory["activities"]:
    findings.extend(privacy_findings(read_utf8(path), path.relative_to(ROOT).as_posix()))
print("Hallazgos por tipo:", Counter(f.kind for f in findings))
print("Documentos que requieren revisión:", len({f.source for f in findings}))
```
